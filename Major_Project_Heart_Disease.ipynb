# **Data Preprocessing and Cleaning**

```
df_enc=pd.get_dummies(df,drop_first=True)
```

```
df_enc
```

```
df_enc=df_enc.astype(int)
```

```
df_enc
```

```
from sklearn.preprocessing import StandardScaler
```

```
numeric_cols=['Age','RestingBP','Cholesterol','MaxHR' ,	'Oldpeak']
scaler=StandardScaler()
df_enc[numeric_cols]=scaler.fit_transform(df_enc[numeric_cols])
```

```
df_enc.head()
```

```
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score,confusion_matrix,f1_score,classification_report
from sklearn.linear_model import LinearRegression
from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import GaussianNB
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
```

```
df_enc.head()
```

```
X=df_enc.drop('HeartDisease',axis=1)
y=df_enc['HeartDisease']
```

```
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)
```

```
models = {
    "LogisticRegression":LogisticRegression(),
    "KNN":KNeighborsClassifier(),
    "SVC":SVC(),
    "Decision TRE":DecisionTreeClassifier(),
    "Naive Bayes":GaussianNB()
}
```

```
result=[]
```

```
# for name ,model in models.items():
#   model.fit(X_train,y_train)
#   y_predict=model.y_predict(X_test)

#   accuracy=accuracy_score(y_test,y_predict)
#   f1=f1_score(y_test,y_predict)
#   result.append({
#       'model':name
#       'accuracy':round(accuracy,4)
#       'F1':round(f1,4)

#   })
for name, model in models.items():
    # Train the model
    model.fit(X_train, y_train)

    # Predict
    y_predict = model.predict(X_test)

    # Evaluate
    accuracy = accuracy_score(y_test, y_predict)
    f1 = f1_score(y_test, y_predict)

    # Store results
    result.append({
        "Model": name,
        "Accuracy": round(accuracy),
        "F1 Score": round(f1)
    })
```
